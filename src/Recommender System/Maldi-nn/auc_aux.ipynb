```python
import numpy as np

log_path = "/export/usuarios01/egarroyo/MALDI_for_AMR_prediction/output.log"

results = {}

current_species = None

with open(log_path, "r") as f:
    for line in f:

        line = line.strip()

        # detectar especie
        if line.startswith("Species:"):
            current_species = line.split("Species:")[1].strip()
            results[current_species] = []
            continue

        # detectar AUC
        if "AUC fold:" in line:
            try:
                auc = float(line.split("AUC fold:")[1].strip())
                results[current_species].append(auc)
            except:
                pass


# ============================================================
# GENERAR TABLA
# ============================================================

print("\n\nFINAL AUC TABLE\n")

print("Species,AUC mean ± std,N folds")

for sp, aucs in results.items():

    if len(aucs) == 0:
        mean_auc = np.nan
        std_auc = np.nan
    else:
        mean_auc = np.mean(aucs)
        std_auc = np.std(aucs)

    print(f"{sp},{mean_auc:.4f} ± {std_auc:.4f},{len(aucs)}")
```
